```python
import pandas as pd
import numpy as np
import os

print("Bibliotecas importadas com sucesso!")
```

```python

raw_path = '../data/raw/datatran2024.csv'

df = pd.read_csv(raw_path, sep=';', encoding='latin1', low_memory=False)

print(f"Dimensões da base carregada: {df.shape[0]} linhas e {df.shape[1]} colunas.")
df.head(3)
```

```python
print("=== 1. VALORES AUSENTES ===")
missing = df.isnull().sum()
print(missing[missing > 0])

print("\n=== 2. VERIFICAÇÃO DE DUPLICIDADE DE LINHAS ===")
duplicadas = df.duplicated().sum()
print(f"Linhas integralmente duplicadas: {duplicadas}")

print("\n=== 3. UNICIDADE DO IDENTIFICADOR (id) ===")
print(f"Total de registros: {len(df)}")
print(f"Valores únicos em 'id': {df['id'].nunique()}")

print("\n=== 4. DISTRIBUIÇÃO DA CLASSIFICAÇÃO DOS ACIDENTES ===")
print(df['classificacao_acidente'].value_counts(dropna=False))

print("\n=== 5. TIPOS DE DADOS ATUAIS PARA CAMPOS CRÍTICOS (km, latitude, longitude) ===")
print(df[['km', 'latitude', 'longitude']].dtypes)
```

```python
df_prep = df.copy()

df_prep['classificacao_acidente'] = df_prep['classificacao_acidente'].fillna('Não Informado')

for col in ['km', 'latitude', 'longitude']:
    if df_prep[col].dtype == 'object':
        df_prep[col] = df_prep[col].astype(str).str.replace(',', '.').astype(float)

print("Tratamentos de limpeza e conversão aplicados com sucesso!")
print(df_prep[['km', 'latitude', 'longitude']].dtypes)
print("Valores ausentes após o tratamento na classificação:", df_prep['classificacao_acidente'].isnull().sum())
```

```python
os.makedirs('../data/processed', exist_ok=True)

output_path = '../data/processed/acidentes_2024_preparados.csv'
df_prep.to_csv(output_path, index=False, sep=';', encoding='utf-8')

print(f"Base de dados preparada salva com sucesso em: {output_path}")
print(f"Dimensões finais: {df_prep.shape[0]} linhas e {df_prep.shape[1]} colunas.")
```
